```python
def eh_palindromo(s):
    """
    Verifica se a string `s` é um palíndromo, ignorando espaços e
    considerando letras maiúsculas e minúsculas como iguais.

    Args:
    s (str): A string a ser verificada.

    Returns:
    bool: True se `s` for um palíndromo, False caso contrário.
    """
    # Remove espaços e transforma a string para minúsculas
    s = s.replace(" ", "").lower()
    # Verifica se a string é igual ao seu reverso
    return s == s[::-1]

# Exemplo de texto para teste
texto = "ANa"  # Altere esta linha para testar diferentes textos

# Verificar se o texto é um palíndromo
if eh_palindromo(texto):
    print(f'O texto "{texto}" é um palíndromo.')
else:
    print(f'O texto "{texto}" não é um palíndromo.')


```
