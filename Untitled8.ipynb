```python
def contar_palavra_na_frase(frase, palavra):
    """
    Conta quantas vezes a palavra `palavra` aparece na frase `frase`.

    Args:
    frase (str): A frase onde a contagem será feita.
    palavra (str): A palavra a ser contada.

    Returns:
    int: O número de vezes que `palavra` aparece na frase.
    """
    # Converte a frase e a palavra para minúsculas para uma contagem case-insensitive
    frase = frase.lower()
    palavra = palavra.lower()

    # Divide a frase em palavras e conta quantas vezes a palavra aparece
    return frase.split().count(palavra)

# Exemplo de uso
if __name__ == "__main__":
    # Solicita a frase e a palavra ao usuário
    frase = input("Digite a frase: ")
    palavra = input("Digite a palavra a ser contada: ")

    # Conta as ocorrências da palavra na frase
    quantidade = contar_palavra_na_frase(frase, palavra)
    print(f"A palavra '{palavra}' aparece {quantidade} vezes na frase.")
```
