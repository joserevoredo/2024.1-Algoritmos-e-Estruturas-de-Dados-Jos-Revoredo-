```python
def inverter_string(s):
    """
    Inverte a string `s`.

    Args:
    s (str): A string a ser invertida.

    Returns:
    str: A string invertida.
    """
    return s[::-1]

# Exemplo de uso
if __name__ == "__main__":
    # Solicita a string ao usuário
    string = input("Digite a string a ser invertida: ")

    # Inverte a string
    string_invertida = inverter_string(string)

    # Exibe o resultado
    print(f"A string invertida é: '{string_invertida}'")
```
