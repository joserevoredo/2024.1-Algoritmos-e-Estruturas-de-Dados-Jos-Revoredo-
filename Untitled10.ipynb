```python
def transformar_em_maiusculas(frase):
    """
    Converte a frase fornecida para maiúsculas.

    Args:
    frase (str): A frase a ser convertida para maiúsculas.

    Returns:
    str: A frase convertida para maiúsculas.
    """
    return frase.upper()

# Exemplo de uso
if __name__ == "__main__":
    # Solicita a frase ao usuário
    frase = input("Digite a frase que deseja converter para maiúsculas: ")

    # Converte a frase para maiúsculas
    frase_maiusculas = transformar_em_maiusculas(frase)

    # Exibe a frase convertida
    print(f"A frase em maiúsculas é: '{frase_maiusculas}'")





```
